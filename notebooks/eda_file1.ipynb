```python
hourly = new_df1.groupby('hour').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

hourly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha tranzaksiyalar soni')
plt.xlabel('Soat')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
hourly['umumiy_hajm'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha umumiy hajm', color='darkorange')
plt.xlabel('Soat')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
weekly = new_df1.groupby('day_of_week').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

kun_tartibi = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekly = weekly.reindex(kun_tartibi)


weekly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Hafta kunlari bo\'yicha tranzaksiyalar soni', color='seagreen')
plt.xlabel('Hafta kuni')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
weekly['umumiy_hajm'].plot(kind='bar', figsize=(12,5), color='pink', title='Hafta kunlari bo\'yicha umumiy hajm')
plt.xlabel('Hafta kuni')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
monthly = new_df1.groupby('month').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)
oy_tartibi = ['January', 'February', 'March', 'April', 'May', 'June', 
              'July', 'August', 'September', 'October', 'November', 'December']
monthly = monthly.reindex(oy_tartibi)

monthly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Oy bo\'yicha tranzaksiyalar soni')
plt.xlabel('Oy')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
## Columnlar bo'yicha foydalanish imkoniyatlari

created_at — bu eng asosiy ustun, band/bo'sh vaqtlarni aniqlash uchun ishlatildi (soat, hafta kuni bo'yicha). Kelajakda oy/fasl bo'yicha sezonlik tahlil qilish uchun ham ishlatish mumkin.

volume, amount, price — savdo hajmi va daromadni hisoblash uchun asosiy ustunlar. Bulardan foydalanib qaysi vaqt/kunda eng ko'p daromad tushishini aniqlash mumkin va narx o'zgarishi savdo hajmiga qanday ta'sir qilganini tekshirish mumkin.

dispenser_id — qaysi kolonka qanchalik band ekanini bilish uchun, shu orqali kolonkalarni tekshirish/ta'mirlash jadvalini rejalashtirish mumkin.

operator_id — operatorlar orasida ish yukini solishtirish uchun, xodimlarni to'g'ri taqsimlash uchun foydali.

car_id — doimiy mijozlarni aniqlash uchun ishlatish mumkin

temperature, density — yoqilg'i sifatini nazorat qilish maqsadida kiritilgan ustunlar, lekin hozirgi holatda ma'lumot sifati past, shuning uchun avval sensorlarni tekshirish kerak, keyingina sifat nazorati uchun ishonchli foydalanish mumkin bo'ladi.

shift_id — smenalar kesimida ish yukini tahlil qilish uchun ishlatish mumkin (created_at bilan birga), lekin asosiy vaqt tahliliga shart emas.

fuel_type_violation, pressure_violation, technically_inspection, without_violations — qoidabuzarliklarni nazorat qilish va hisobot berish uchun ishlatilishi mumkin.

branch_id, district_id, region_id — bitta fayl doirasida konstant bo'lgani uchun foydasiz, lekin barcha 5 ta filialni birlashtirib solishtirishda kerak bo'ladi.

arm_price, rnn, status, is_send, tenant_id — bu faylda hech qanday
```
