```python
new_df2 = df2[["id", "created_at", "filial_created_at", "price", "volume",
               "amount", "dispenser_id", "operator_id",
               "car_id", "temperature", "density", "branch_id",
               "shift_id", "arm_price"]].copy()
```

```python
new_df2["filial_created_at"] = pd.to_datetime(new_df2["filial_created_at"], format='mixed')
```

```python
new_df2["created_at"] = pd.to_datetime(new_df2["created_at"])
```

```python
new_df2["filial_created_at"].dtype
```

```python
new_df2["created_at"].dtype
```

```python
new_df2['local_time'] = new_df2['filial_created_at'] + pd.Timedelta(hours=5)
```

```python
new_df2["hour"] = new_df2["local_time"].dt.hour
```

```python
new_df2[['local_time', 'hour']].head()
```

```python
new_df2["day_of_week"] = new_df2["local_time"].dt.day_name()
```

```python
new_df2['month'] = new_df2['local_time'].dt.month_name()
```

```python
new_df2[['local_time', 'hour', 'day_of_week', 'month']].head()
```

```python
hourly = new_df2.groupby('hour').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

hourly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha tranzaksiyalar soni')
plt.xlabel('Soat')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
hourly
```

```python
hourly['umumiy_hajm'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha umumiy hajm', color='darkorange')
plt.xlabel('Soat')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
weekly = new_df2.groupby('day_of_week').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

kun_tartibi = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekly = weekly.reindex(kun_tartibi)


weekly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Hafta kunlari bo\'yicha tranzaksiyalar soni', color='seagreen')
plt.xlabel('Hafta kuni')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
weekly
```

```python
weekly['umumiy_hajm'].plot(kind='bar', figsize=(12,5), color='pink', title='Hafta kunlari bo\'yicha umumiy hajm')
plt.xlabel('Hafta kuni')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
monthly = new_df2.groupby('month').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)
oy_tartibi = ['January', 'February', 'March', 'April', 'May', 'June', 
              'July', 'August', 'September', 'October', 'November', 'December']
monthly = monthly.reindex(oy_tartibi)

monthly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Oy bo\'yicha tranzaksiyalar soni')
plt.xlabel('Oy')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
monthly['umumiy_hajm'].plot(kind='bar', figsize=(12,5), color='pink', title='Oylar bo\'yicha umumiy hajm')
plt.xlabel('Oy nomi')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

## Note: **dispenser_id**

Barcha 10 ta kolonka taxminan bir xil davr (7+ oy) davomida ishlagan bo'lsada, kolonka 9 va 10 boshqalariga qaraganda 10+ marta kamroq tranzaksiyaga ega (1664 va 1123 ta, boshqalarida 11500-54000 ta). 
Bu farq ishlash muddati bilan bog'liq emas, shuning uchun sababi texnik nosozlik, joylashuv yoki maxsus foydalanish bo'lishi mumkin.

```python
new_df2['hisoblangan_amount'] = new_df2['price'] * new_df2['volume']
farq = new_df2['amount'] - new_df2['hisoblangan_amount']
farq.describe()
```

```python
new_df2.loc[farq.idxmin()]
```

```python
new_df2[farq.abs() > 1000].shape
```

```python
new_df2[farq.abs() > 1000][['id', 'created_at', 'price', 'volume', 'amount', 'hisoblangan_amount']]
```

```python
new_df2["volume"].count()
```

```python
new_df2[new_df2["amount"] <= 0].shape
```

```python
new_df2[new_df2['volume'] <= 0]['volume'].value_counts().head(10)
```

```python
new_df2_clean = new_df2[(new_df2['volume'] > 0) & (farq.abs() <= 1000)].copy()
```

```python
print("Asl qatorlar soni:", len(new_df2))
print("Tozalangandan keyin:", len(new_df2_clean))
print("Chiqarib tashlangan:", len(new_df2) - len(new_df2_clean))
```

```python
new_df2_clean['hisoblangan_amount_v2'] = new_df2_clean['price'] * new_df2_clean['volume']
farq_v2 = new_df2_clean['amount'] - new_df2_clean['hisoblangan_amount_v2']
farq_v2.describe()
```
