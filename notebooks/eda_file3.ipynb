```python
new_df3 = df3[["id", "created_at", "filial_created_at", "price", "volume",
               "amount", "dispenser_id", "operator_id",
               "car_id", "temperature", "density", "branch_id",
               "shift_id", "arm_price"]].copy()
```

```python
new_df3["filial_created_at"] = pd.to_datetime(new_df3["filial_created_at"], format='mixed')
```

```python
new_df3["created_at"] = pd.to_datetime(new_df3["created_at"], format="mixed")
```

```python
new_df3["created_at"].dtype
```

```python
new_df3['local_time'] = new_df3['filial_created_at'] + pd.Timedelta(hours=5)
```

```python
new_df3["hour"] = new_df3["local_time"].dt.hour
```

```python
new_df3[['local_time', 'hour']].head()
```

```python
new_df3["day_of_week"] = new_df3["local_time"].dt.day_name()
```

```python
new_df3['month'] = new_df3['local_time'].dt.month_name()
```

```python
new_df3[['local_time', 'hour', 'day_of_week', 'month']].head()
```

```python
hourly = new_df3.groupby('hour').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

hourly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha tranzaksiyalar soni')
plt.xlabel('Soat')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
hourly
```

```python
hourly['umumiy_hajm'].plot( kind='bar', figsize=(12,5), title='Soat bo\'yicha umumiy hajm', color='darkorange')
plt.xlabel('Soat')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
weekly = new_df3.groupby('day_of_week').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)

kun_tartibi = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekly = weekly.reindex(kun_tartibi)


weekly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Hafta kunlari bo\'yicha tranzaksiyalar soni', color='seagreen')
plt.xlabel('Hafta kuni')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
weekly
```

```python
weekly['umumiy_hajm'].plot(kind='bar', figsize=(12,5), color='pink', title='Hafta kunlari bo\'yicha umumiy hajm')
plt.xlabel('Hafta kuni')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

```python
monthly = new_df3.groupby('month').agg(
    tranzaksiyalar_soni=('id', 'count'),
    umumiy_hajm=('volume', 'sum')
)
oy_tartibi = ['January', 'February', 'March', 'April', 'May', 'June', 
              'July', 'August', 'September', 'October', 'November', 'December']
monthly = monthly.reindex(oy_tartibi)

monthly['tranzaksiyalar_soni'].plot( kind='bar', figsize=(12,5), title='Oy bo\'yicha tranzaksiyalar soni')
plt.xlabel('Oy')
plt.ylabel('Tranzaksiyalar soni')
plt.show()
```

```python
monthly['umumiy_hajm'].plot(kind='bar', figsize=(12,5), color='pink', title='Oylar bo\'yicha umumiy hajm')
plt.xlabel('Oy nomi')
plt.ylabel('Umumiy hajm (litr)')
plt.show()
```

## Note: **dispenser_id**

Barcha 8 ta kolonka taxminan bir xil davr (11+ oy) davomida ishlagan bo'lsada, kolonka 1 va 2 boshqalariga qaraganda ozgina kamroq tranzaksiyaga ega. 
Bu farq ishlash muddati bilan bog'liq emas, shuning uchun sababi texnik nosozlik, joylashuv yoki maxsus foydalanish bo'lishi mumkin.

```python
new_df3['hisoblangan_amount'] = new_df3['price'] * new_df3['volume']
farq = new_df3['amount'] - new_df3['hisoblangan_amount']
farq.describe()
```

```python
new_df3.loc[farq.idxmin()]
```

```python
new_df3[farq.abs() > 1000].shape
```

```python
new_df3[farq.abs() > 1000][['id', 'created_at', 'price', 'volume', 'amount', 'hisoblangan_amount']]
```

```python
new_df3[new_df3["volume"] <= 0].shape
```

```python
new_df3["volume"].count()
```

```python
new_df3[new_df3['volume'] <= 0]['volume'].value_counts().head(10)
```

```python
new_df3_clean = new_df3[(new_df3['volume'] > 0) & (farq.abs() <= 1000)].copy()
```

```python
print("Asl qatorlar soni:", len(new_df3))
print("Tozalangandan keyin:", len(new_df3_clean))
print("Chiqarib tashlangan:", len(new_df3) - len(new_df3_clean))
```

```python
new_df3_clean.nlargest(10, 'volume')[['id', 'dispenser_id', 'volume', 'amount']]
```

```python
new_df3_clean[new_df3_clean['volume'] <= 0].shape
new_df3_clean[new_df3_clean['amount'] <= 0].shape
```

```python
new_df3_clean['hisoblangan_amount_v2'] = new_df3_clean['price'] * new_df3_clean['volume']
farq_v2 = new_df3_clean['amount'] - new_df3_clean['hisoblangan_amount_v2']
farq_v2.describe()
```

```python
kerakli_ustunlar = ['id', 'local_time','hour', 'day_of_week', 'month', 'dispenser_id', 'price', 'volume', 'amount', 'branch_id']
new_df3_final = new_df3_clean[kerakli_ustunlar].copy()
```
